```python
from bip_utils import Bip39MnemonicGenerator, Bip39SeedGenerator, Bip84, Bip84Coins, Bip44Changes

# 1. Generate a new 12-word seed phrase
mnemonic = Bip39MnemonicGenerator().FromWordsNumber(12)
print(f"Seed Phrase: {mnemonic}")

# 2. Convert the seed phrase into a seed (raw bytes)
seed_bytes = Bip39SeedGenerator(mnemonic).Generate()

# 3. Initialize a BIP84 Native SegWit wallet (modern Bitcoin format)
bip84_master = Bip84.FromSeed(seed_bytes, Bip84Coins.BITCOIN)

# 4. Derive the first receiving account (Derivation Path: m/84'/0'/0'/0/0)
# Purpose(84) -> Coin(0 for BTC) -> Account(0) -> Change(0 for receiving) -> AddressIndex(0)
wallet = bip84_master.Purpose().Coin().Account(0).Change(Bip44Changes.CHAIN_EXT).AddressIndex(0)

# 5. Extract keys and address
print(f"Private Key (WIF): {wallet.PrivateKey().ToWif()}")
print(f"Bitcoin Address: {wallet.PublicKey().ToAddress()}")
```
